# Summary and Next Steps

## What You Built in This Course

<div style="background: #F9F7F4; border-radius: 10px; padding: 20px 24px; box-shadow: 0 2px 8px rgba(27,49,57,0.06); border-top: 6px solid #1B5162; margin: 16px 0;">
<div style="font-size: 14pt; color: #0b2026; line-height: 1.7;">

Over the course of these lessons, you went from an empty workspace to a fully automated data engineering pipeline. Specifically, you:

- **Got oriented** in the Databricks workspace and Unity Catalog.
- **Created and queried** Unity Catalog tables using SQL.
- **Modified** table data with `INSERT`, `UPDATE`, and `DELETE`.
- **Used time travel** to query previous versions of a UC table.
- **Ingested data** with multiple methods: `CTAS`, the Upload UI, and `COPY INTO`.
- **Built a Medallion Architecture pipeline** that progressively refines data through Bronze, Silver, and Gold layers.
- **Automated a pipeline** with a Lakeflow Job that runs your transformation work end-to-end.
- **(Bonus)** Wrote a **Spark Declarative Pipeline** that lets the platform manage orchestration, dependencies, and quality for you.

These are the same building blocks data engineers use in production Databricks workspaces every day.

## Continue Building With the Genie Code

<div style="background: #F9F7F4; border-radius: 10px; padding: 20px 24px; box-shadow: 0 2px 8px rgba(27,49,57,0.06); border-top: 6px solid #FF5F46; margin: 16px 0;">
<div style="font-size: 14pt; color: #0b2026; line-height: 1.7;">

The **Genie Code** lives in every notebook and SQL editor. You can use it to write SQL or Python from natural language, explain code you don't understand, debug error messages, and get suggestions as you type.

Two ways to invoke it:

- Click the **Genie icon** in the right sidebar to open the chat panel.
- Press **Cmd + I** (Mac) or **Ctrl + I** (Windows) inside any cell to invoke the in-cell Genie Code prompt.

Use it to keep building beyond this course — describe what you want to do in plain language and let Genie Code help draft the code.
