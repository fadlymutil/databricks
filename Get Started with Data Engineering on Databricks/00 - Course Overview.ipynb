# Get Started with Databricks for Data Engineering

## Overview

This course introduces the foundational skills needed to perform a basic data engineering workflow on the **Databricks Data Intelligence Platform**. You will explore the workspace, work with Unity Catalog, and learn the day-to-day building blocks a data engineer uses on Databricks.

The course follows a hands-on path. You start by getting oriented in the workspace, then work through paired Demo and Lab notebooks for each topic. Demos walk you through a concept with the instructor or a guided notebook; Labs ask you to apply what you just saw on your own.

The content covers:

- The Databricks Data Intelligence Platform and how the Databricks workspace, Unity Catalog, and notebooks fit together.
- Creating and managing **Delta Lake** tables.
- Modifying data with `INSERT`, `UPDATE`, and `DELETE`.
- Exploring **version history** and **time travel** on UC tables.
- **Ingesting** data with **Lakeflow Connect** options: `CTAS`, the Upload UI, and `COPY INTO`.
- Building a **Medallion Architecture** pipeline that transforms data through Bronze, Silver, and Gold layers.
- **Automating** pipelines with **Lakeflow Jobs**.
- **(Bonus)** Building a declarative pipeline with **Apache Spark™ Declarative Pipelines**.

By the end of the course, you will be comfortable creating Delta Lake tables, ingesting data into them, transforming that data through a Medallion pipeline, and automating the whole thing as a scheduled job.

## Terminal Objectives

By the end of this course, you will be able to:

- **Describe** how the Databricks Data Intelligence Platform unifies data and AI on a single governed foundation.
- **Navigate** the Databricks workspace, Unity Catalog, and notebooks to locate and work with data.
- **Create** and **modify** Delta Lake tables using SQL, and **query** previous versions using time travel.
- **Ingest** data into Delta Lake using multiple methods, including `CTAS`, the Upload UI, and `COPY INTO`.
- **Build** a Medallion Architecture pipeline that transforms data through Bronze, Silver, and Gold layers.
- **Automate** a data pipeline using a Lakeflow Job.
- **(Bonus) Build** a declarative pipeline with **Apache Spark™ Declarative Pipelines**.

##### This Academy Edition is based on course version 4.0.3.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **Basic familiarity with cloud data platforms.** You understand what a database, table, and query are at a conceptual level.
- **Working knowledge of SQL.** You can read and write `SELECT` statements with `WHERE`, `GROUP BY`, and aggregates, and you understand `INSERT`, `UPDATE`, and `DELETE`.
- **General awareness of data engineering concepts** like data ingestion, transformation, and pipelines is helpful but not required — the course covers the basics.

### B1. Your Course Catalog

<div style="border-left: 4px solid #1976d2; background: #e3f2fd; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">

#### Catalog Information

- The setup at the top of each Demo and Lab **creates a Unity Catalog catalog and schema for you** the first time it runs, then reuses them.
- The **Create Catalog** permission is required.
- The catalog name is derived from your Databricks username and follows this pattern: **`learn_get_started_de_YOUR_USERNAME`**, where `YOUR_USERNAME` is the part of your sign-in email before `@`, lowercased, with each run of characters other than letters and digits replaced by `_`, and limited to 19 characters.
- The course schema (**get_started_de**) and a **myfiles** volume holding the course CSV files are created inside that catalog by the setup script.

</div>
