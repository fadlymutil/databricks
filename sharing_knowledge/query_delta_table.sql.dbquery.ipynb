```sql
select current_catalog(), current_schema();

create database if not exists workspace.dbsimas;


use workspace.dbsimas;

create table customer (cif string, name string, insert_date timestamp);

describe customer;

describe extended customer;


describe history customer;

insert into customer values ('001','Ardhan dan dan', current_timestamp()), ('002','theo widi jago banget', current_timestamp());

select * from customer;


select * from customer timestamp as of ('2026-10-30T08:38:44.000+00:00');

select * from customer@v0;

restore table customer to version as of 0;

select * from customer;

describe history customer;

restore table customer to version as of 1;


create table customer_iceberg (cif string, name string, insert_date timestamp)
using iceberg;


describe extended customer_iceberg;


select * from customer 
union 
select * from customer_iceberg;

create table customer_uniform (cif string, name string, insert_date timestamp)
USING DELTA
TBLPROPERTIES (
  'delta.universalFormat.enabledFormats' = 'iceberg',
  'delta.enableIcebergCompatV2' = 'true'
);

describe extended customer_uniform;


```
