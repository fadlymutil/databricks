##### EXPAND FOR ADDITIONAL NOTES

<details>

### How Unity Catalog addresses your data

- The whole visual is about one idea: **where does a table live, and how do you name it.**
- Traditional SQL used a **two-level** namespace, `database.table`. You had a database, and objects inside it. That was the deepest you could go.
- Unity Catalog adds a **third level on top**, the **catalog**. So the full address is now `catalog.schema.object`.

### The three levels, top to bottom

- **Catalog** — the top-tier container. Think of it as the biggest bucket you organize data into.
- **Schema** — sits inside a catalog and groups related objects together. (You may hear "schema" and "database" used interchangeably; in Unity Catalog they are the same thing.)
- **Object** — the thing you actually query or manage: a table, view, volume, function, or model.

The nested boxes in the visual show this literally: the object lives inside the schema, and the schema lives inside the catalog. You reference any object by its full three-part path, for example `sales.orders.transactions`.

### Why the extra level helps

- A catalog gives teams a clean way to carve up data at the highest level. Common patterns:
  - **By business unit** — separate sales, marketing, and HR data.
  - **By environment** — keep dev, staging, and prod apart (SDLC).
  - **As sandboxes** — a throwaway space for temporary datasets.
- You can have as many catalogs as you want, and each can hold as many schemas as you want.

### Writing less with USE

- Typing `catalog.schema.object` every time gets tedious. The `USE` statements set a default catalog and schema:
  - `USE CATALOG sales;`
  - `USE SCHEMA orders;`
- After that, you can reference an object by its name alone (for example `transactions`), and Databricks fills in the rest. You only need the full path when you want something outside your current defaults.

</details>

### C4. Creating and Managing Views

1. Let's create a **vw_gold** view that presents an aggregated version of the **patient_info** table data by averaging heart rate data per patient on a daily basis. 

    In the following cell, we will:
    - Create the view
    - Query the view

```sql
-- Create a gold view
CREATE OR REPLACE VIEW vw_gold AS (
  SELECT 
    mrn, 
    name, 
    MEAN(heartrate) AS avg_heartrate, 
    DATE_TRUNC("DD", time) AS date
  FROM patient_info
  WHERE name IS NOT NULL
  GROUP BY mrn, name, DATE_TRUNC("DD", time)
);


-- View the rows inside the gold view
SELECT * 
FROM vw_gold;
```

```sql
%python
# Python equivalent (creates a temp view)

from pyspark.sql import functions as F

# Build the gold aggregation with the DataFrame API
gold_df = (
    spark.table("patient_info")
    .where(F.col("name").isNotNull())
    .groupBy("mrn", "name", F.date_trunc("DD", F.col("time")).alias("date"))
    .agg(F.mean("heartrate").alias("avg_heartrate"))
    .select("mrn", "name", "avg_heartrate", "date")
)

# Register it as a temporary view (session-scoped, not a persistent Unity Catalog view)
gold_df.createOrReplaceTempView("vw_gold_py")

# View the rows
display(gold_df)
```
