# Lab - Catalog Explorer Navigation
### Estimated Duration: ~7-10 minutes
In this demo, you inspect data objects in Unity Catalog using the Catalog Explorer. Once you can navigate the Catalog Explorer, you can find any dataset, see who owns it, and trace how it is used.

## Learning objectives

By the end of this lesson, you should be able to:

- **Navigate** the Catalog Explorer to locate catalogs, schemas, and tables in Unity Catalog.
- **Review** and manage object details, including permissions and metadata.
- **Add** AI-generated descriptions and column comments to improve data discovery.
- **Visualize** data lineage at both the table and column levels.
- **Query** tables using natural language through Genie in the Catalog Explorer.
- **Explore** table history, insights, and quality monitoring features.

## B. The Catalog Explorer

All of the data objects you create are available to you in the Catalog Explorer. You can view and change permissions, change object ownership, examine lineage, and a whole lot more.

## C. Add AI-Generated Descriptions and Comments

When you open a table in Catalog Explorer, Databricks automatically generates a suggested description using AI. Review it for accuracy before accepting. The description of a table (or other object) is used in many places throughout your Lakehouse.

- **Discovery and search** - Descriptions make tables easier to find. When you search for data in Catalog Explorer or use AI-powered search, Databricks indexes descriptions to help match your intent with relevant tables.

- **AI-powered features** - Genie Agents and AI assistants use table and column descriptions to understand your data semantics. When you ask questions in natural language, these descriptions help the AI generate accurate queries and provide better recommendations.

- **Documentation and collaboration** - Descriptions serve as inline documentation for your team. Anyone browsing the Catalog Explorer can quickly understand what a table contains and how it should be used, without needing to examine the data or contact the table owner.

- **Governance and compliance** - Descriptions can indicate data sensitivity, usage policies, or compliance requirements. Combined with tags, they help data stewards understand what data exists and how it should be handled.

1. Select **Overview** to go back to the main **patient_info** table.

2. Accept the suggested table description by clicking **Accept**.

    > Beyond table descriptions, column comments help users understand individual fields, their formats, and expected values. With the button labeled **AI Generate**, Databricks can generate comments for your columns.

3. In the column details section select **AI Generate**.

    The Data Intelligence Platform proposes comments for each column in your table.

4. Click **Save all** to accept all suggested comments.

**NOTE:** Always review generated AI output. For training, we will leave as is.

## D. Understand Data Lineage

Data lineage in Databricks tracks the complete lifecycle of your data through Unity Catalog, showing where data originates, how it transforms, and where it is consumed. It captures dependencies at both the table and column level, mapping relationships between tables, views, notebooks, jobs, dashboards, pipelines, and ML models.

### D1. View Data Lineage in Catalog Explorer

1. Click the **Lineage** tab to see upstream and downstream dependencies listed.

2. Click **See lineage graph** to open an interactive visual representation showing:
    - Upstream sources (where data originates)
    - Downstream consumers (where data is used)
    - Connected notebooks, jobs, dashboards, and pipelines

3. Explore column-level lineage: click any column name to see which source columns map to it.

4. Close the lineage graph by clicking the **X** in the upper-right corner.

## E. Query a Table Using Natural Language

You can query a table directly, using natural language, from within the Catalog Explorer. This is done using Genie.

1. Click the **Sample data** tab.

    This tab shows the first rows of the table, so you can see the data it contains. You can also query the table using Genie.

2. Enter the following query:

    > What is the average heart rate per device?

    Genie shows you the SQL it generated, so you can review it before running.

3. Click **Yes** to allow Genie to run the query.

    After the query runs, you will see the results displayed right in the Catalog Explorer.

## F. Explore Other Table Information

There are other tabs along the top where you can view and manage metadata about this table:

- **Overview** - On the right, you have information about the table, tags, its description, and you can add a row filtering function if you wish. On the left, you get information about each column in the table.

- **Sample data** - This tab shows the first rows of the table.

- **Details** - You get the same information here as you would by running `DESCRIBE EXTENDED <table_name>`.

- **Permissions** - The UI gives you the ability to manage table permissions. You can `GRANT` and `REVOKE` permissions here.

- **History** - The Delta Lake transaction log is on this tab. You can get this programmatically by running `DESCRIBE HISTORY <table_name>`.

- **Lineage** - Provides a list and a graph of upstream and downstream dependencies.

- **Insights** - The Databricks Data Intelligence Platform provides these insights, so you can see how your data object is being used.

- **Quality** - This tab gives you the ability to monitor this table for data quality.
