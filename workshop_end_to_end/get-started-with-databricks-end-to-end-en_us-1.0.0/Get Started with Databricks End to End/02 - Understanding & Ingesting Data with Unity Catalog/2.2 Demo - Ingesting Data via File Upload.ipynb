# Demo - Ingesting Data via File Upload

One of the simplest ways to bring data into Databricks is to upload it directly from your local machine. In this lesson, you upload a CSV file, create a Unity Catalog table from it, and explore the result.

## Learning objectives

By the end of this lesson, you should be able to:

- **Upload** a CSV file from your local machine into Databricks.
- **Create** a Unity Catalog table from the uploaded file.
- **Configure** column names and data types during the upload.
- **Explore** the table using both the Catalog UI and SQL.

## B. Download the CSV File

In the same folder as this notebook, there is a data file named **cust_service_data.csv**. Download it so you can upload it in the next section.

1. If it is not already open, click the ![Workspace Icon](https://files.training.databricks.com/binder/prod_main/get-started-with-databricks-end-to-end-en_us-1.0.0/images/20260922T145855Z/Get Started with Databricks End to End/Includes/images/workspace-icon.png "Workspace Icon") **Workspace** icon just to the right of the left navigation bar.

2. Hover over the **cust_service_data.csv** file and click the three-dot (kebab) menu to the right of the file name.

3. Click **Download**.

## C. Create a Table from the File

Now create a table by uploading the file you just downloaded.

1. Right-click **Catalog** in the left navigation bar and select **Open Link in New Tab**. 
    - This keeps this notebook open while you work.

2. In the new tab, open your **labuser_UNIQUE_ID** catalog, then the **example** schema.

3. In the upper-right corner, click **Create** > **Table**.

4. Drag the **cust_service_data.csv** file you downloaded onto the drop zone.

5. Click the downward-facing arrow to the right of the **interaction_id** column name and select **Exclude column**. 
    - The column is grayed out and will not be included in the table.
    
6. Name the table **cust_service_data**.

7. Click **Create table**.
