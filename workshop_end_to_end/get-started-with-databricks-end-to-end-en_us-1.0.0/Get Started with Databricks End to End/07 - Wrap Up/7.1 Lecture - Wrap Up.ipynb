# Databricks End-to-End: Course wrap up

In this lesson, we'll recap the main workflows and point to resources for continuing your learning.

## Key workflows recap

Throughout this course, you've practiced the core workflows across Databricks: workspace navigation, data ingestion, notebooks, SQL, dashboards, and AI features. Let's review the core competencies you've developed.

### Navigating the Databricks workspace

You've learned how to efficiently navigate and organize your work in the Databricks environment:

* **Workspace Explorer**: Browse folders, notebooks, and files in your workspace hierarchy
* **Creating and organizing assets**: Set up logical folder structures for projects and teams
* **Search functionality**: Quickly locate notebooks, queries, dashboards, and tables
* **Collaboration**: Share notebooks and collaborate with team members

### Ingesting data

You've learned to upload local CSV files to a Unity Catalog volume and create a table from them.

### Building and running notebooks

Notebooks are the heart of interactive development in Databricks:

* **Cell execution**: Run individual cells or entire notebooks
* **Visualizations**: Create charts and graphs directly from DataFrames and SQL results
* **Markdown documentation**: Document your analysis with rich text, images, and links
* **Version control**: Integrate with Git for version management and collaboration

### Running SQL queries

Databricks SQL provides a powerful environment for analytics and reporting:

* **SQL Editor**: Write and execute SQL queries with autocomplete and syntax highlighting
* **Query history**: Review and rerun past queries
* **Data exploration**: Use SQL to explore schemas, tables, and data quality

### Building dashboards

Transform queries into actionable insights with interactive dashboards:

* **Dashboard creation**: Build visual dashboards from SQL queries and notebook results
* **Visualization widgets**: Add charts, tables, counters, and filters to your dashboards
* **Layout customization**: Arrange widgets in a grid layout for optimal presentation
* **Sharing and permissions**: Share dashboards with stakeholders and control access levels
* **AI/BI dashboards**: Create natural language-powered dashboards with Genie

### AI/ML and Genie

Leverage AI-powered tools to accelerate your analytics and machine learning workflows:

* **Genie**: Ask natural language questions about your data and get instant answers with automatically generated queries and visualizations
* **AI-assisted coding**: Get code suggestions and completions while writing Python, SQL, and other languages
* **ML workflows**: Build, train, and deploy machine learning models using MLflow integration
* **Genie Code for data science**: Automatically train and tune models with minimal configuration
