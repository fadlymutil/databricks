# Building Dashboards from Queries

## Overview

In this demonstration, you build an interactive AI/BI Dashboard on the `samples.bakehouse` data. You create a new dashboard, use **Genie Code** with a natural language prompt to generate the visualizations, review the datasets Genie Code creates, and publish and share the finished dashboard.

## Learning Objectives

By the end of this lesson, you will be able to:

1. **Create** a new Databricks AI/BI Dashboard.
2. **Use** Genie Code with a natural language prompt to generate dashboard visualizations.
3. **Publish** a dashboard and manage sharing permissions for workspace and account users.

## A. Using the Sample Data

You use the `bakehouse` sample dataset in the `samples` catalog. This dataset simulates a bakery franchise business with sales transactions, customer data, and franchise information.

**Dataset location:**

- **Catalog**: `samples`
- **Schema**: `bakehouse`

This lesson uses the following tables as the source data for the dashboard visualizations:

- **sales_transactions**: transaction-level sales data including products, prices, quantities, and payment information.
- **sales_customers**: customer demographic and contact information.

## B. Create a New Dashboard

1. Right-click **Dashboards** in the left navigation bar and open the link in a new tab.

2. In the new tab, click **Create dashboard** in the upper-right corner.

3. At the top of the screen, click the dashboard name and change it to **FIRST_NAME_LAST_INITIALS_Bakehouse Dashboard**.

## D. Review the Dashboard Data

With a brand-new dashboard, you normally associate the dashboard with data before designing the display, but Genie Code has already done this for you. Take a look at the data sources it created.

At the top of the dashboard screen, you have two tabs, **Data** and **Untitled page**.

- **Data**: defines the datasets used in the dashboard. Datasets are bundled with dashboards when you share, import, or export them using the UI or API.

- **Untitled page**: the dashboard canvas where you create visualizations and construct your dashboard. Each item on the canvas is called a **widget**. Widgets come in three types: visualizations, text boxes, and filters.

1. You start on the **Untitled page** tab. 

2. Select the **Data** tab to see the data sources.

3. Here you can see the data sources **Genie Code** used for the dashboard.

4. Click their names in the left column, under **Datasets**, to view their contents, measures, and dimensions. You can add additional datasets if you wish.

## E. Publishing and Sharing

When your dashboard is complete, you can publish it and share it with others. When you publish a dashboard, the default setting is **Share data permission**.

  border-left: 4px solid #1976d2;

  <strong style="display:block; color:#0d47a1; margin-bottom:6px; font-size: 1.1em;">

  <div style="color:#333;">

- **Share data permission**: viewers use your data permissions to access the data and power the queries behind the dashboard. This enables a shared cache, which can improve performance.
- **Individual data permission**: each viewer uses their own data permissions to run queries. This can lead to more frequent refresh operations, and viewers need access to both the underlying tables and the SQL warehouse to see the visualizations.

  </div>
</div>

To publish your dashboard, complete the following steps:

1. Click **Publish** in the upper-right corner of your dashboard.

2. In the **Publish** dialog, you will see two data permission options. Select **Share data permission**.

3. Optionally, notify viewers via email and describe your changes in the text field.

4. Click **Publish** in the lower-right corner of the dialog. The **Sharing** dialog opens afterward. If it does not open, select **Share** next to **Publish** at the top of the dashboard.

   - Use the text field to search for individual users, or share the dashboard with a preconfigured group like **Admins** or **All workspace users**. From this window, you can grant leveled privileges like **Can Manage** or **Can Edit**. See <a href="https://docs.databricks.com/en/security/auth-authz/access-control/index.html#lakeview" target="_blank">Dashboard ACLs</a> for details on permissions.
   - The bottom of the **Sharing** dialog controls view access. Use this setting to share with all account users.

5. Under **Sharing settings**, choose **Anyone in my account can view** from the drop-down, then close the **Sharing** dialog. This makes the dashboard visible to every user in your Databricks account. For sensitive data, use **Specific users** instead.

6. Use the context icon at the top of the dashboard, next to the dashboard name, to switch between **Edit Draft** and **View Published** versions of your dashboard.
