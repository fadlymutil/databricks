# Lab: Create a visualization for your Bakehouse dashboard
### Estimated Duration: ~5-10 minutes

## Learning objectives

In this lab, you'll practice creating a new visualization and adding it to the **Bakehouse Dashboard** you created in the previous lesson. This hands-on exercise will reinforce your understanding of:

* Creating datasets from SQL queries
* Configuring visualizations
* Adding visualizations to an existing dashboard
* Formatting and styling charts

## B. Position Your widget and publish

### Position and resize

Organize your dashboard for the best viewing experience:

1. **Move the widget**: Click and drag the visualization to reposition it
   * Try placing it near your other sales-related visualizations
   * Consider visual flow - group related metrics together

2. **Resize the widget**: Drag the edges or corners to adjust the size
   * Make it large enough to read comfortably
   * But not so large that it dominates the dashboard

3. **Rearrange other widgets** if needed to create a balanced layout

### Test your visualization

Before republishing, verify everything works correctly:
1. Click **Refresh** on the counter. 
1. Confirm the number matches your earlier query result. 

### Republish your dashboard

Now that the counter is ready, republish the dashboard to make the changes live:

1. Click the **Publish** button in the top right corner
1. Click **Publish** to make your changes live
1. Click the published version (**View published**) and confirm the new counter shows up.

## C. View the AI/BI Dashboard in Genie One

1. Open Genie One in a new tab.

2. Select **Dashboards** from the left navigation pane.

3. Browse the dashboards available to you.
   - You should see the dashboard you published.
   - Depending on your workspace and the permissions granted on other published dashboards, you may see additional dashboards.
