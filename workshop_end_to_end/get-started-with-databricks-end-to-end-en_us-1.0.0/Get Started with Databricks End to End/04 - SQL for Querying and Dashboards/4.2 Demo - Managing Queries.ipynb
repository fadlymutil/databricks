# Managing Queries

## Overview

In this demonstration, you organize the queries you save in the Databricks SQL Editor. You save a query to the workspace, arrange your saved queries into folders, tag them, and use query history to find and revisit previous runs.

All examples use the `samples.bakehouse` dataset.

## Learning Objectives

By the end of this lesson, you will be able to:

1. **Save** a query to the workspace and give it a descriptive name.
2. **Organize** queries into folders and apply tags.
3. **Use** query history to review and recover past runs.

## A. Create and Save a Query

You start by writing a query, running it, and saving it to the workspace so you can find it again later.

1. In the far left sidebar, right-click **SQL Editor** and open the link in a new tab.

2. Create a new query.

3. In the SQL Editor, set the query context at the top:

   - **Catalog**: `samples`
   - **Schema**: `bakehouse`

4. Copy the query below and paste it into the editor.

## B. Organizing Queries

By default, queries are saved in your **Home** folder. Folders and tags help you and your team find them as the number of saved queries grows.

### B1. Create Folders for Organization

1. In the left sidebar, right-click **Workspace** and open the link in a new tab.

2. Click **Home** in the upper-left. You will see the query you saved above.

3. Click the **Create** button in the upper-right and select **Folder**.

4. Name the folder, for example `Bakehouse Analytics`. You are taken into the new folder.

5. Go back to your Home folder by clicking **Home** in the upper-left.

6. Drag and drop your saved query into the folder.

7. You can create nested folders for deeper organization, similar to any computer's file system:

   - `Marketing Reports` then `Customer Segmentation`
   - `Sales Analytics` then `Daily Reports`, `Monthly Reports`

### B2. Add Tags to Queries

1. Open your saved query.

2. Click the **three dots (⋮)** menu at the top of the query editor. 

3. Select the (⋮) > **File** > **Edit query tags**.

4. Save tags such as:

   - `customer-analytics`
   - `sales`
   - `monthly-report`
   - `executive-dashboard`

## C. Query History

Query history lets you revisit past runs of a query without re-executing it.

### C1. Access Query History

1. Open and run any saved query in the SQL Editor.

2. In the upper-right of the results section, click the **Open Execution History** icon to see information about past runs.

    ![History](https://files.training.databricks.com/binder/prod_main/get-started-with-databricks-end-to-end-en_us-1.0.0/images/20260922T145855Z/Get Started with Databricks End to End/Includes/images/query-execution-history.png)

### C2. Review Past Query Results

1. In the Query History panel, click any past execution.

2. From there you can:

   - **View the results** from that specific run.
   - **See the exact query** that was executed, which is useful if the query has since been modified.
   - **Download historical results** from past runs.
   - **Check execution details** such as duration and rows returned.
