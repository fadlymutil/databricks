# Creating and Running Queries

In this demonstration, you write SQL queries using the Databricks SQL Editor.

## Learning objectives

By the end of this lesson, you should be able to:

- **Open** the SQL Editor.
- **Run** a query against the bakehouse data.
- **Save** and rerun a query.

## A. The Databricks SQL Editor

The SQL Editor is a dedicated environment for writing and running SQL queries. Use it when you want a focused SQL workspace, separate from notebooks. You might prefer this editor, especially if you have experience with one in another SQL development environment. In Databricks, the SQL Editor gives you a more developer-focused experience with specific features.

## B. Open and Use the SQL Editor

### B1. Opening and using the SQL editor

1. In the far left sidebar menu, right click on **SQL Editor** and open the link in a new tab. This will take you to the SQL editor.

2. Click the **SQL Query** button under the *Create New* section.
    - Or click the **plus icon (`+`)** next to *New tab* and select **New Query** to start a fresh SQL editor session.

3. Copy the directions below and paste them into the query editor. Then follow those instructions.
