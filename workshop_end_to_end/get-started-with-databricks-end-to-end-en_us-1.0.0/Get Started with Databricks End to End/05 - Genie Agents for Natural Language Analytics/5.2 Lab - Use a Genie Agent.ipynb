# Lab - Use a Genie Agent
### Estimated Duration: ~20-30 minutes


## Overview

In this short hands-on lab, you use a **Genie Agent** to analyze the `bakehouse` data using natural language. You ask a question, refine it for a better answer, and then ask a more powerful question that spans multiple tables. This lab takes about 10 minutes.

## Learning Objectives

By the end of this lab, you will be able to:

1. **Ask** a business question in a Genie Agent and interpret the result.
2. **Refine** a question to get a more precise answer.
3. **Ask** a question that Genie answers by joining multiple tables.

## A. Ask a Basic Question

Open your **Bakehouse** Genie Agent and ask the following questions, one at a time. After each answer, click **Show code** to see the SQL Genie generated.

1. Ask a simple aggregation:

   > *What is the total revenue?*

   You should get a single number ($66,471). Notice how Genie sums the totalPrice column.

2. Break the total down by category:

   > *Show me revenue by product*

   Genie groups by product and returns one row per product, listing all 6 bakehouse products.

## B. Refine for Better Results

Your first question does not always give you exactly what you want. Refining it makes Genie's response more precise.

1. Start with a broad request:

   > *Show me sales*

   This is generic and returns raw transaction records, which are hard to scan.

2. Refine it to be specific about aggregation and sorting:

   > *Show me total sales by product, ordered from highest to lowest*

   Compare the two results. The refined question returns summarized, sorted data with the top products first, which is much easier to act on.

## C. Ask a More Powerful Question

Genie remembers context within a conversation and can join tables for you automatically.

1. Ask a question, then a one of the follow-up questions below that depends on it:

   > *Which product has the highest revenue?*

   > *Show me all transactions for that product*

   Genie remembers which product you meant from the previous question and filters to it, without you naming it again.

2. Ask a question that needs two tables:

   > *Show me transactions with customer names*

   Genie joins **sales_transactions** to **sales_customers** on its own. You did not have to specify the join. Click **Show code** to confirm the join in the generated SQL.
