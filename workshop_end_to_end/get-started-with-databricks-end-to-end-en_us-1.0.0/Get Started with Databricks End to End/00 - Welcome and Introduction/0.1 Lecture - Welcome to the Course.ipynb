# Welcome to Databricks End-to-End

## Your complete introduction to the Databricks Data Intelligence Platform

This course is a comprehensive, hands-on introduction to the Databricks Data Intelligence Platform. You will follow a single end-to-end journey that starts in the workspace and finishes with an introduction to AI and machine learning, covering the everyday skills you need to work with data on Databricks.

The course covers:

- Navigating the Databricks workspace, compute options, and the Lakehouse versus Genie experiences.
- Understanding and ingesting data with Unity Catalog, including file upload and the Catalog Explorer.
- Working in Databricks notebooks and using Genie Code to write and debug code.
- Querying data with SQL, managing saved queries, and building dashboards and visualizations.
- Exploring data in natural language with Genie Agents.
- An introduction to AI and machine learning with MLflow and the Data Science Agent.

By the end of the course, you will be comfortable navigating Databricks, finding and governing data in Unity Catalog, writing SQL and notebook code with AI assistance, building dashboards, and exploring data with Genie.

## Course Objectives

By the end of this course, you will be able to:

- **Navigate** the Databricks workspace, compute options, and core UI.
- **Explain** how Unity Catalog governs data through its three-level namespace, and **ingest** data using file upload and the Catalog Explorer.
- **Use** Databricks notebooks and **apply** Genie Code to generate and debug code.
- **Write** SQL queries, **manage** saved queries, and **build** dashboards and visualizations.
- **Explore** data in natural language using Genie Agents.
- **Describe** how MLflow and the Data Science Agent support AI and machine learning workflows on Databricks.

## A. Prerequisites

This is an introductory course. To get the most out of it, you should have:

- **Basic computer and web browser skills.** You can navigate a web application and work with tabs and menus.
- **A willingness to work with data.** No prior Databricks experience is required.
- **Helpful, but optional: basic SQL.** Familiarity with `SELECT` statements makes the SQL and dashboard modules easier, but the course introduces what you need as you go.

## B. The Sample Dataset

Throughout the course, we will be using the Bakehouse dataset. the Bakehouse dataset is a curated sample bakery franchise business dataset, built around sales transactions and related business data. There are five tables in this dataset:
1. sales_transactions - Core sales data
2. sales_franchises - Store locations
3. sales_customers - Customer information
4. sales_suppliers - Supplier data
5. media_customer_reviews - Customer feedback

This dataset is perfect for learning about retail analytics, geographic analysis, customer behavior, sales trends, and multi-table joins. Many of these topics will be covered in this course.
