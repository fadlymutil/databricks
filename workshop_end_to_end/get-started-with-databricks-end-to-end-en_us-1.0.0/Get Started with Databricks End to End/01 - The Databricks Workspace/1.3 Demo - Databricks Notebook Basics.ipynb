# Demo - Databricks Notebook Basics

A **notebook** is where you write and run code in Databricks. It is made up of **cells**: code cells run code, and markdown cells hold formatted text like these instructions. In this lesson you will get to know the notebook interface, run Python and SQL, and document your work with markdown.

## Learning objectives

By the end of this lesson, you should be able to:

- **Navigate** the notebook interface and its sidebar tools.
- **Attach** a notebook to compute so you can run code.
- **Run** code cells in Python and SQL and view their output.
- **Create** and **edit** markdown cells, and **re-order** cells.

## C. Work with Code Cells

Code cells run code, and the output appears directly below the cell.

### C1. Run a Python cell

1. Click the Python cell below to select it.
2. Run it with the **Run** (&#9654;) icon in the cell's upper-left corner, or press **Shift + Enter**.
3. Note the output that appears below the cell.

```python
# This Python cell runs on the compute you attached.
print("Hello World")

print("2 + 2 =", 2 + 2)
```

### C2. Override the cell language with SQL

As you saw in the toolbar, this notebook's default language is **Python**. You can still run another language in a single cell using a **magic command**. Starting a cell with **`%sql`** runs SQL.

1. Run the SQL cell below (**Shift + Enter**) and note the output.

```python
%sql
SELECT "Hello World" AS greeting, 2 + 2 AS simple_math
```

### C3. Add a cell and run all

- **Add a cell:** hover just below a cell and click the **(+) Code** icon to insert a new code cell. Try typing `print("I added this cell")` and running it.
- **To run everything:** click **Run all** at the top of the notebook to run every cell in order.

## D. Work with Markdown Cells

Markdown cells add formatted text and documentation to your notebook, like the instructions you are reading now.

- **Create one:** hover above or below a cell and click the **(+) Text** icon. A new cell is added with `%md` at the top, which designates it as markdown.
- **Edit one:** double-click a markdown cell to edit it. Press **`Esc`** or run the cell to render it.

Use the next cell to get to know markdown better.

### D1. Markdown example

Double-click this cell to see the markdown behind it, then press **`Esc`** to render it again.

# Title one
## Title two
### Title three

This is text with a **bold** word in it.

This is text with an *italicized* word in it.
