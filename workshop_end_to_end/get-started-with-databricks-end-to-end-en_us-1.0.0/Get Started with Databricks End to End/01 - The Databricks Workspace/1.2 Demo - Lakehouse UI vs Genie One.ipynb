# Demo - Lakehouse UI vs Genie One

Databricks has several experiences, but the two most people use for data and analytics are the full **Lakehouse workspace UI** you have been touring and **Genie One**, a simplified interface built for business users. This lesson explains the difference, who sees which one, and how to move between them.

## Learning objectives

By the end of this lesson, you should be able to:

- **Explain** what Genie One is and how it differs from the Lakehouse workspace UI.
- **Switch** between the Lakehouse workspace UI and Genie One using the app switcher.
- **Navigate** the Genie One home page, **open** shared assets, and **ask** a question in natural language.
- **Describe** how entitlements determine which interface a user sees.

## B. What Is Genie One?

Genie One is a simplified, business-user-friendly interface that provides a single, intuitive entry point to interact with data and AI in Databricks. 

Unlike the full Lakehouse workspace UI, Genie One allows users to focus on what matters most: getting access to their data quickly and efficiently. 

Genie One streamlines the Databricks experience with a chat interface that can access your data, and a search interface to help you find the assets that you need when you need them.

<details>

### What Genie One Brings Together

- Genie One is the simplified Databricks UI for business users. It is a single entry point for working with data and AI, without having to navigate technical concepts like compute, queries, models, or notebooks.
- The visual is a containment metaphor on purpose. The big outer box is Genie One, and everything a consumer needs lives inside it. The message is "one home for all of this," not "a tool that links to other tools."

### The three assets inside

- **AI/BI Dashboards** — curated charts and KPIs. Business users view and interact with them to track key performance indicators and analyze metrics.
- **Genie Agents** — ask questions about your data in natural language and get answers back, no SQL required.
- **Databricks Apps** — interactive, purpose-built analytics and workflow applications, including custom apps your teams build.

### All in one place, you can

- **Ask questions in natural language** — a unified, full-screen chat that spans your dashboards and agents, with follow-ups and root-cause analysis, no SQL required.
- **Search assets by name** to jump straight to what you need, and favorite or browse them by domain and business context.
- **Schedule chat tasks** so recurring questions run on their own and post results back as a thread, with email notifications when they run.
- **Draft and share documents** straight from a chat conversation, edit them in a canvas pane, share by link, and track version history.
- **Take action with agents and apps** — move from insight to action with Genie Agents and Databricks Apps, no code required.
- **Work anywhere** — chat and view dashboards from the iOS and Android app, or natively in Slack and Microsoft Teams with @mentions.
- **Trust every answer** — responses respect Unity Catalog row- and column-level security and source permissions, and you can verify the data behind an answer with citations.

### A couple of deeper capabilities

- **Connect 50+ external tools** — pull in context from sources like Google Drive, SharePoint, GitHub, Glean, and Atlassian.
- **Citations** — click the citation icons in a response to verify the knowledge sources and underlying data.

### Who it is for

- Designed for business users, who need at minimum Consumer access.
- Works at the workspace level for a single workspace, or at the account level for a unified view across all workspaces in your Databricks account.
- When you open an asset it launches in the familiar experience for that object type, and whether it opens read-only or with an Edit draft button depends on your privileges, covered next.

</details>


### B2. Opening and Interacting with Assets in Genie One

1. Click the app switcher icon (![App Switcher](https://files.training.databricks.com/binder/prod_main/get-started-with-databricks-end-to-end-en_us-1.0.0/images/20260922T145855Z/Get Started with Databricks End to End/Includes/images/app-switcher.png)).

2. Right click on **Genie One** and select **Open Link in New Tab**.

3. The Genie One home page is the central hub for business users. It is organized into several key areas:

    - **Search bar** allows you to do two things:
        1. **Search** to quickly locate assets
        2. **Ask** ask questions about your data

    - **Quick links** (in the left panel), a hub for assets such as dashboards, Genie Agents, and apps.

    **The Genie One interface is an AI-powered chat over your whole data estate that queries all the data you have access to and finds answers to your questions.**

4. For example, select **Dashboards**.
    - Depending on your workspace, you might see dashboards if available.

<div style="border-left: 4px solid #4299E0; background: rgba(66,153,224,0.10); padding: 14px 18px; border-radius: 4px; margin: 16px 0; font-family: sans-serif;">
  <strong style="display:block; color:#2272B4; margin-bottom:6px; font-size:15pt;">Opening an asset in Genie One</strong>
  <div style="color:#0b2026; font-size:14pt; line-height:1.55;">

When you click an asset in Genie One, it opens in the familiar experience for that object type (for example, a dashboard opens in the dashboard viewer).

An asset opens in different ways depending on whether you have edit privileges on it:

- If you have **viewing rights only**, the asset opens within Genie One.
- If you have **authoring privileges**, you will see an **Edit draft** button to open the asset in the full Lakehouse workspace UI for editing.

</div>

### B3. Ask Genie One a Question

**Chat** in Genie One is a unified, full-screen, natural-language interface where business users can ask questions about their data. 

In this step, you will ask Genie One a question about the **samples.nyctaxi.trips** sample data and watch it find the answer for you.

1. Select **+ New Chat**.

2. Ask **Genie One** the question below. Click **Copy**, then paste it into the chat and send it.
    - Every Databricks workspace comes with a set of ready-to-use sample tables in the **samples** catalog, so you can try Chat right away, no setup or SQL required.
