### C2. Load the Data into a DataFrame

Run the cell below. It loads the sales table into a DataFrame variable and previews it with `display()`. The `sales_df` variable stays available in the cells that follow.

```python
# Load the sales data into a DataFrame and preview it
sales_df = spark.table("samples.bakehouse.sales_transactions")
display(sales_df.limit(20))
```

### C3. Aggregate and Visualize

1. Run the cell below to calculate total revenue by product.

2. Then build a chart from the result:

   - Click the **+** icon above the results table and select **Visualization**.
   - Choose a **Bar** chart.
   - Set the **X column** to `product` and the **Y column** to `total_revenue`.

**On your own:**

- Add a **markdown cell** (`%md`) below the chart with a short heading and a bullet or two describing what you found.

- Click **Run all** at the top to run the whole notebook top to bottom.

```python
from pyspark.sql import functions as F

# Total revenue per product, highest first
product_revenue = (
    sales_df.groupBy("product")
    .agg(F.sum("totalPrice").alias("total_revenue"))
    .orderBy(F.desc("total_revenue"))
)

display(product_revenue)
```
