# Demo - Navigating the Databricks Workspace

This notebook is a guided tour of the Databricks Workspace UI: the homepage, your user settings, the left sidebar, search and the **+ New** menu, how to organize your assets, and the different types of compute. A few things look different in Databricks Free Edition, and we will flag those as we go.

## Learning objectives

By the end of this lesson, you should be able to:

- **Navigate** the Databricks Workspace homepage.
- **Access** and **configure** your user settings.
- **Use** the left sidebar to find features across SQL, data engineering, and AI/ML.
- **Use** intelligent search, **Recents**, and the **+ New** menu to find and create assets.
- **Organize** workspace assets with folders and files.
- **Identify** the different types of Databricks compute.

## A. The Workspace Homepage

1. Right click the **Databricks logo** at the top left and select **Open Link in a New Tab**
   - You'll be taken to a screen that says **Welcome to Databricks**.

2. This page contains some helpful resources:
   - **Suggested**
      - These are objects you've recently opened or worked on, including notebooks, data tables, pipelines, and more.
   - **Favorites**
      - This is a list of objects you have "favorited" by selecting the Star icon next to its name.

3. Select **Workspace** in the main navigation bar to get your user files.

4. Close the browser tab.

## B. Locate and Configure User Settings

1. At the top right, click your **user icon** (a letter from your username) and select **Settings**.

2. **Settings** opens to **User Settings**, which has five sections:

   - **Profile**: your display name and the group(s) you belong to.
   - **Preferences**: **Language** (General) and **Interface Theme** (Appearance). Change these if you wish.
   - **Developer**: **Access Tokens**, **SQL Query Snippets**, and **Editor Settings** (General, Code Editor, Experimental Features, SQL Format).
   - **Linked Accounts**: Git integration. Link a Git account and select a **Personal Access Token** for a provider such as GitHub.
   - **Notifications**: manage **Model Registry**, **Comment**, **Account-level**, and **Promotional** email notifications.

**NOTE:** Admins also see a **Workspace admin** menu. Those options are not covered in this content.

## D. Search and the "+ New" Button

### D1. Intelligent Search

Search helps you quickly find notebooks, queries, dashboards, tables, and more across your workspace. It is fuzzy, so partial names and keywords still surface the right results.

**Finding assets through the search bar:**
- Click the **search bar** at the top-center (or press **Cmd/Ctrl + P**).
- Type the name of a table or notebook.
- Select a result from the dropdown, or press **Enter** to see all matches.
- Use filters to narrow results by type, such as **Notebooks**, **Dashboards**, or **Tables**.

**Finding recent files and folders:**
- On the left sidebar, right click on **Recents** and open in a new tab.
- Select the file, notebook, or table you want to revisit.

For more, see the **Search for workspace objects** docs: [AWS](https://docs.databricks.com/aws/en/search/) | [Azure](https://learn.microsoft.com/en-us/azure/databricks/search/) | [GCP](https://docs.databricks.com/gcp/en/search/).

### D2. The '+ New' Menu

**+ New** at the top of the sidebar is the quickest way to create something. The options a new user reaches for most are:

- **Notebook** - Interactive document for writing and running code in Python, SQL, Scala, or R. (Scala and R are not available in Databricks Free Edition.)
- **Query** - A SQL query you can save, share, and schedule.
- **Dashboard** - A visual report of charts, tables, and insights from your data.
- **Genie Agent** - Ask questions about your data in natural language.
- **Job** - An automated workflow that runs notebooks, queries, or code on a schedule.

The menu also has more advanced options such as **ETL Pipeline**, **Experiment**, **Model**, **App**, and (under **More**) **File**, **Git folder**, **SQL Warehouse**, and **Serving Endpoint**. **Cluster** (under **More**) is not available in Databricks Free Edition.

var rcCurrent = null;
function rcSelect(id) {
  var wrap = document.getElementById('rc-detail-wrap');
  var card = document.getElementById('rc-detail-card');
  var d = REVEAL[id];
  document.querySelectorAll('.rc-box').forEach(function(b) {
    b.classList.toggle('active', parseInt(b.dataset.id) === id && rcCurrent !== id);
