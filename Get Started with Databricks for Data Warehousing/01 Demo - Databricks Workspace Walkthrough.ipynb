### 3. Tour the workspace sidebar

The left sidebar is your home base in Databricks. Hover over the **hamburger icon** (☰) at the top-left to expand it.

The items you will use most often in this course:

| Sidebar item | What it's for |
|---|---|
| **+ New** | Create a notebook, query, dashboard, job, or pipeline |
| **Workspace** | Browse and organize your own notebooks and folders |
| **Catalog** | Browse governed data: catalogs, schemas, tables, and volumes |
| **SQL Editor** | Author and save SQL queries that run on a SQL Warehouse |
| **Dashboards** | Build AI/BI dashboards from your data |
| **Query History** | See every query that has run, with timings and a profile |
| **SQL Warehouses** | View and configure SQL-optimized compute |

### 4. Find data in the catalog

All governed data in Databricks lives in **Unity Catalog**, organized in a three-level hierarchy: **catalog → schema → table**. Let's find a sample table that's available in every Databricks workspace.

#### Browse to the table

1. In the left sidebar, click **Catalog** to open Catalog Explorer.
1. Find and expand the **`samples`** catalog.
1. Expand the **`nyctaxi`** schema.
1. Click the **`trips`** table.
1. On the right panel, look at the **Overview** tab to see the column list, then click the **Sample Data** tab to preview a few rows.

You are looking at NYC taxi trip data: pickup and dropoff times, fare amounts, distances, and zip codes.

#### Query that same table from this notebook

Run the cell below to query the table you just previewed.

```python
%sql
SELECT *
FROM samples.nyctaxi.trips
LIMIT 10;
```

You are now seeing the same data you previewed in Catalog Explorer, just queried directly in the notebook. This is the loop you will use throughout the course: **find data in the catalog, then work with it in a notebook.**

### 5. Notebook basics: cells, languages, and Genie Code
Databricks notebooks support multiple languages in the same notebook. The cell below is Python. It counts the rows in the sample table and prints the result.

```python
trip_count = spark.sql("SELECT COUNT(*) AS total FROM samples.nyctaxi.trips").collect()[0]["total"]
print(f"There are {trip_count:,} trips in the sample dataset.")
```

#### Changing cell type

Try adding a new cell to this notebook:

1. Hover between this cell and the next one. A **+ Code** button appears. Click it to add a new cell below.
1. In the new cell, set the language to **SQL** using the dropdown at the top-right of the cell (or type `%sql` on the first line).
1. Type: `SELECT MAX(fare_amount) AS biggest_fare FROM samples.nyctaxi.trips`
1. Press **Shift + Enter** to run it.

#### Try Genie Code

Genie Code can write, explain, and debug code for you. There are two ways to open it:

- Click the **Genie icon** in the right sidebar to open the chat panel.
- Press **Cmd + I** (Mac) or **Ctrl + I** (Windows) inside a cell to invoke the in-cell Genie Code prompt.

Try asking it: *"Write a SQL query that counts trips by pickup zip code from samples.nyctaxi.trips, sorted by trip count descending."*

### 6. SQL Warehouses
