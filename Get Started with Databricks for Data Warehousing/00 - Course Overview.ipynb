# Get Started with Databricks for Data Warehousing

## Overview

This course introduces the foundational skills a warehousing practitioner needs to do day-to-day data warehousing work on the **Databricks Data Intelligence Platform**. You will explore the workspace, query Unity Catalog managed tables, optimize them for analytical reads, load data incrementally, model a star schema, deliver insights through dashboards and external BI tools, and diagnose slow queries, all from a single, governed environment.

The course follows a hands-on path. After getting oriented in the workspace, you will work through guided **Demo** notebooks for each topic that walk you through key concepts with worked examples.

The content covers:

- The **Databricks Data Intelligence Platform** and how the workspace, **Unity Catalog**, and **SQL Warehouses** fit together.
- Writing SQL queries against governed Unity Catalog tables and visualizing the results inline.
- Keeping **UC managed tables** fast for analytical reads with **Predictive Optimization** and **automatic liquid clustering** (`CLUSTER BY AUTO`), and where manual commands still fit.
- Loading data into the warehouse incrementally using **Auto Loader**, with safe, exactly-once re-runs.
- Modeling a **star schema** in Unity Catalog with fact and dimension tables, surrogate keys, and informational primary and foreign key constraints.
- Defining a **Unity Catalog metric view** so every consumer (SQL Editor, dashboards, Genie Agents, alerts, and external BI tools) reads one trusted set of measure definitions.

By the end of the course, you will be comfortable querying governed Unity Catalog tables, keeping UC managed tables optimized for analytical workloads, loading data incrementally with Auto Loader, modeling a star schema, and delivering consistent metrics through metric views.

## Terminal Objectives

By the end of this course, you will be able to:

- **Navigate** the Databricks workspace, Unity Catalog, and SQL Warehouses to locate tables, run queries, and inspect results.
- **Write** SQL queries that join, aggregate, and filter Unity Catalog tables, and turn the results into inline visualizations.
- **Keep** UC managed tables fast for analytical reads with Predictive Optimization and automatic liquid clustering (`CLUSTER BY AUTO`), and recognize when manual commands such as `OPTIMIZE` or `ANALYZE` still apply.
- **Load** data into UC managed tables incrementally with Auto Loader, including verifying that re-runs do not duplicate rows.
- **Build** a star schema in Unity Catalog with fact and dimension tables, surrogate keys, a procedurally-generated date dimension, and informational primary and foreign key constraints.
- **Define** a Unity Catalog metric view and explain how the same definition feeds the SQL Editor, dashboards, Genie Agents, alerts, and external BI tools.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **Basic familiarity with the data warehousing domain.** You understand at a conceptual level what a data warehouse is, what a table is, and what business intelligence (BI) tools are used for.
- **Working knowledge of SQL.** You can read and write `SELECT` statements with `WHERE`, `GROUP BY`, aggregate functions, and `JOIN`.
- **No prior Databricks knowledge is assumed.** Databricks-specific concepts (Unity Catalog, Delta Lake, SQL Warehouses, metric views) are introduced as they come up.

## D. The Vintage Audio Co. Dataset

Every demo and lab in this course uses a single dataset that we built specifically for warehousing scenarios: **Vintage Audio Co.**, a fictional retailer that sells vintage hi-fi audio equipment through 30 stores across the United States.

The dataset is shipped as CSV files and prepared in your workspace automatically by each lesson's setup. The assets the course uses:

| Asset | Created where | Used by |
|---|---|---|
| `customers` table | `vintage_audio` schema | Lessons 02, 04 |
| `products` table | `vintage_audio` schema | Lessons 02, 03, 04, 05 |
| `stores` table | `vintage_audio` schema | Lessons 02, 03 |
| Six quarterly `orders_*.csv` files | `data` volume | Lessons 06, 07 (Auto Loader ingestion) |

The order CSV files are intentionally left in the volume rather than pre-loaded as tables, because loading them with Auto Loader is the subject of Lesson 06.
